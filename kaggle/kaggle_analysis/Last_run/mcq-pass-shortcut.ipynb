````python
"""
OpenRouter Pilot 1 v2 - Adversarial Distractor Generation + Blind Shortcut Audit

Tests 5 research-backed prompt variants for generating Strategy-2-style distractors
(paraphrased wrong meanings) and audits them with a 4-model OpenRouter committee.

Research bases:
- Zengaffinen et al. 2026: solve-first + misconception-based distractors; revealing
  the correct answer improves quality.
- Bitew et al. 2023: few-shot predictive prompting with retrieved examples.
- Haladyna / VERGE taxonomy: partial info, contextual misalignment, critical dependency.
- MoMentS: blind options-only audit as the evaluation.
- AlgGeoTest: overgenerate-then-select.
- NoisyBench: iterative filtering; distractors must not contain gold or alter stem.

Outputs:
    pilot1_generated_mcqs.csv      -- all generated MCQs by prompt variant
    pilot1_audit_results.csv       -- raw blind-audit votes
    pilot1_prompt_comparison.csv   -- per-variant consensus accuracy
    pilot1_summary.json            -- full run summary + cost history
"""

import os
import re
import json
import time
import random
import hashlib
import requests
import numpy as np
import pandas as pd
from collections import Counter
from pathlib import Path
from datetime import datetime, timezone

# ── Configuration ───────────────────────────────────────────────────────────

N_PER_LANG = 5
SEED = 20260615
COST_CAP_USD = 5.0
TEMPERATURE = 0.0
MAX_TOKENS_GEN = 600
MAX_TOKENS_AUDIT = 8

OPENROUTER_URL = "https://openrouter.ai/api/v1/chat/completions"

# Generator model: GPT-5 is the strongest OpenAI model available on OpenRouter
# as of 2026-06-15 and is cheaper than Opus 4.8, keeping the comparison under $5.
# (Claude Fable 5 launched June 9 but was suspended by Anthropic on June 12 for
# export-control compliance, so it is not a reliable choice right now.)
GENERATOR_MODEL = "openai/gpt-5"

AUDIT_COMMITTEE = [
    # Model selection verified against OpenRouter's live catalog on 2026-06-15.
    # - Opus 4.9 does not exist / has not been announced.
    # - Gemini 3.5 Pro is announced but not yet available on OpenRouter.
    # - Gemini 2.5 Pro is the latest Google Pro model actually on OpenRouter.
    "anthropic/claude-opus-4.8",              # strongest reliable Anthropic model
    "openai/gpt-5",                            # strong & cost-effective
    "google/gemini-2.5-pro",                   # latest Gemini Pro on OpenRouter
    "nvidia/nemotron-3-ultra-550b-a55b:free",  # free tier, useful tie-breaker
]

# OpenRouter pass-through pricing (USD per 1M tokens) as of 2026-06-15.
PRICE_TABLE = {
    "anthropic/claude-opus-4.8": {"input": 5.0, "output": 25.0},
    "openai/gpt-5": {"input": 1.25, "output": 10.0},
    "google/gemini-2.5-pro": {"input": 1.25, "output": 10.0},
    "nvidia/nemotron-3-ultra-550b-a55b:free": {"input": 0.0, "output": 0.0},
}
DEFAULT_PRICE = {"input": 2.0, "output": 8.0}

# Paths
# Hardcoded Kaggle dataset path. Falls back to local cleaned data only when running outside Kaggle.
DATA_DIR = "/kaggle/input/datasets/abrahamsunday123/full-data-complete"
OUTPUT_DIR = "/kaggle/working/"
if not os.path.exists(DATA_DIR):
    DATA_DIR = "actual_data/cleaned"
    OUTPUT_DIR = "openrouter_pilot1_output"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# API key
OPENROUTER_API_KEY = None

try:
    from kaggle_secrets import UserSecretsClient
    OPENROUTER_API_KEY = UserSecretsClient().get_secret("OPENROUTER_API_KEY")
except Exception:
    pass

if not OPENROUTER_API_KEY:
    OPENROUTER_API_KEY = os.environ.get("OPENROUTER_API_KEY")

if not OPENROUTER_API_KEY:
    raise ValueError(
        "OPENROUTER_API_KEY not found. Add it to Kaggle Secrets or set it as an environment variable."
    )

# ── Prompt variants ─────────────────────────────────────────────────────────

# Load from JSON. On Kaggle, upload pilot1_prompt_variants.json to one of the
# paths below (e.g., as a dataset mounted at /kaggle/input/proverbgap-prompts/).

def load_prompt_variants():
    candidate_paths = [
        "/kaggle/input/proverbgap-prompts/pilot1_prompt_variants.json",
        "/kaggle/input/proverbgap-data/pilot1_prompt_variants.json",
        "pilot1_prompt_variants.json",
    ]
    for path in candidate_paths:
        if os.path.exists(path):
            with open(path, encoding="utf-8") as f:
                data = json.load(f)
            print(f"Loaded prompt variants from {path}")
            return data["prompt_variants"]
    raise FileNotFoundError(
        "pilot1_prompt_variants.json not found. Upload it to Kaggle input or place it in the working directory."
    )

PROMPT_VARIANTS = load_prompt_variants()

# ── Cost tracking ───────────────────────────────────────────────────────────

class CostTracker:
    def __init__(self, cap):
        self.cap = cap
        self.spent = 0.0
        self.history = []

    def estimate(self, model, input_tokens, output_tokens):
        p = PRICE_TABLE.get(model, DEFAULT_PRICE)
        return (input_tokens * p["input"] + output_tokens * p["output"]) / 1e6

    def would_exceed(self, model, input_tokens, output_tokens):
        return (self.spent + self.estimate(model, input_tokens, output_tokens)) > self.cap

    def add(self, model, input_tokens, output_tokens, purpose):
        p = PRICE_TABLE.get(model, DEFAULT_PRICE)
        cost = (input_tokens * p["input"] + output_tokens * p["output"]) / 1e6
        self.spent += cost
        self.history.append({
            "model": model,
            "purpose": purpose,
            "input_tokens": input_tokens,
            "output_tokens": output_tokens,
            "cost_usd": cost,
            "cumulative_usd": self.spent,
        })
        return cost

    def check(self):
        if self.spent >= self.cap:
            raise RuntimeError(
                f"Cost cap ${self.cap:.2f} exceeded (spent ${self.spent:.2f}). Aborting."
            )

cost_tracker = CostTracker(COST_CAP_USD)


# ── Cross-platform safe printing ────────────────────────────────────────────

def safe_print(*args, **kwargs):
    try:
        print(*args, **kwargs)
    except UnicodeEncodeError:
        try:
            text = " ".join(str(a) for a in args)
            print(text.encode("ascii", "ignore").decode("ascii"), **kwargs)
        except Exception:
            pass

# ── Data loading ────────────────────────────────────────────────────────────

def load_data():
    paths = {
        "English": os.path.join(DATA_DIR, "English_cleaned.csv"),
        "Arabic": os.path.join(DATA_DIR, "Arabic_cleaned.csv"),
        "Yoruba": os.path.join(DATA_DIR, "Yoruba_cleaned.csv"),
    }
    records = []
    for lang, path in paths.items():
        df = pd.read_csv(path, encoding="utf-8")
        if lang == "English":
            df = df.rename(columns={"Proverb": "proverb", "Correct_Meaning": "meaning"})
        elif lang == "Arabic":
            df = df.rename(columns={"source_text": "proverb", "english_translation": "meaning"})
        else:
            df = df.rename(columns={"Source_Text_Yo": "proverb", "Target_Text_En": "meaning"})
        df = df[["proverb", "meaning"]].dropna()
        df["language"] = lang
        df["sample_id"] = df.index.astype(str)
        records.append(df)
    return pd.concat(records, ignore_index=True)


def sample_per_language(df, n, seed):
    rng = np.random.default_rng(seed)
    sampled = []
    for lang, group in df.groupby("language"):
        group = group.reset_index(drop=True)
        k = min(n, len(group))
        if k == len(group):
            sampled.append(group)
        else:
            idx = rng.choice(len(group), size=k, replace=False)
            sampled.append(group.iloc[idx])
    return pd.concat(sampled, ignore_index=True)


# ── OpenRouter client ───────────────────────────────────────────────────────

def openrouter_chat(model, messages, max_tokens, retries=3):
    headers = {
        "Authorization": f"Bearer {OPENROUTER_API_KEY}",
        "Content-Type": "application/json",
        "HTTP-Referer": "https://www.kaggle.com/",
        "X-Title": "ProverbGap Pilot 1 v2",
    }
    payload = {
        "model": model,
        "messages": messages,
        "temperature": TEMPERATURE,
        "max_tokens": max_tokens,
    }
    for attempt in range(retries):
        try:
            r = requests.post(OPENROUTER_URL, headers=headers, json=payload, timeout=120)
            r.raise_for_status()
            return r.json()
        except Exception as e:
            if attempt == retries - 1:
                raise
            time.sleep(2 ** attempt)
    return None


def extract_text(response):
    if response and "choices" in response and len(response["choices"]) > 0:
        return response["choices"][0]["message"].get("content", "")
    return ""


def extract_usage(response):
    usage = response.get("usage", {}) if response else {}
    return usage.get("prompt_tokens", 0), usage.get("completion_tokens", 0)


# ── JSON parsing ────────────────────────────────────────────────────────────

REASONING_TAGS = re.compile(r"<think>.*?</think>|<reasoning>.*?</reasoning>", re.DOTALL | re.IGNORECASE)


def clean_model_output(text):
    if not text:
        return text
    text = REASONING_TAGS.sub("", text)
    code_block = re.search(r"```(?:json)?\s*(\[.*?\])\s*```", text, re.DOTALL)
    if code_block:
        return code_block.group(1).strip()
    arr = re.search(r"(\[.*\])", text, re.DOTALL)
    if arr:
        return arr.group(1).strip()
    return text.strip()


def parse_options(text):
    cleaned = clean_model_output(text)
    try:
        opts = json.loads(cleaned)
    except Exception:
        try:
            opts = json.loads(cleaned.strip().strip(",").strip())
        except Exception:
            return None
    if isinstance(opts, list) and len(opts) == 4 and all(isinstance(o, str) for o in opts):
        return [o.strip() for o in opts]
    return None


# ── Validation & fallback ───────────────────────────────────────────────────

def jaccard(a, b):
    """Token-overlap similarity; 0 = distinct, 1 = identical."""
    sa, sb = set(a.lower().split()), set(b.lower().split())
    if not sa and not sb:
        return 1.0
    if not sa or not sb:
        return 0.0
    return len(sa & sb) / len(sa | sb)


def passes_length_check(candidate, reference, threshold=0.35):
    """Character-count parity. Word counts are fragile for Arabic/Yoruba."""
    ref_len = len(reference)
    cand_len = len(candidate)
    if ref_len > 0 and abs(cand_len - ref_len) / ref_len <= threshold:
        return True, "length"
    return False, f"length_ratio={cand_len / max(ref_len, 1):.2f}"


def negative_sample(reference_meaning, all_meanings, rng, n=3):
    others = [m for m in all_meanings if m != reference_meaning]
    if len(others) < n:
        others = others * (n // max(len(others), 1) + 1)
    rng.shuffle(others)
    return others[:n]


def semantically_distinct_negative_sample(reference_meaning, all_meanings, rng, n=1, max_jaccard=0.5):
    """
    Return n fallback distractors that are not identical to the reference and
    have low token overlap with it, reducing the chance of accidentally creating
    a semantically similar near-correct option.
    """
    others = [m for m in all_meanings if m != reference_meaning]
    # Prefer meanings with low Jaccard overlap to the reference.
    low_overlap = [m for m in others if jaccard(reference_meaning, m) <= max_jaccard]
    pool = low_overlap if len(low_overlap) >= n else others
    if len(pool) < n:
        pool = pool * (n // max(len(pool), 1) + 1)
    rng.shuffle(pool)
    return pool[:n]


RAW_OUTPUTS = []


def generate_options(proverb, meaning, language, variant, rng, all_meanings):
    """Generate 4 options using a specific prompt variant."""
    cost_tracker.check()
    system_prompt = variant["system_prompt"]
    user_prompt = variant["user_prompt_template"].format(
        proverb=proverb, language=language, correct_meaning=meaning
    )
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt},
    ]
    prompt_tokens_est = sum(len(m["content"].split()) for m in messages) + 50

    if cost_tracker.would_exceed(GENERATOR_MODEL, prompt_tokens_est, MAX_TOKENS_GEN):
        raise RuntimeError(
            f"Call would exceed ${COST_CAP_USD:.2f} cap (spent ${cost_tracker.spent:.4f}). Skipping generation."
        )

    raw_text = None
    try:
        resp = openrouter_chat(GENERATOR_MODEL, messages, MAX_TOKENS_GEN)
        text = extract_text(resp)
        raw_text = text
        prompt_tok, out_tok = extract_usage(resp)
        cost_tracker.add(GENERATOR_MODEL, prompt_tok or prompt_tokens_est, out_tok, f"generation-{variant['name']}")
        opts = parse_options(text)
        if opts is None:
            raise ValueError("Could not parse JSON options.")
    except Exception as e:
        # Hard fallback to semantically distinct negative sampling
        distractors = semantically_distinct_negative_sample(meaning, all_meanings, rng, n=3)
        opts = [meaning] + distractors
        status_meta = {"status": "parse_fallback", "error": str(e)}
        RAW_OUTPUTS.append({
            "stage": "generation",
            "variant": variant["name"],
            "language": language,
            "proverb": proverb,
            "correct_meaning": meaning,
            "raw_model_output": raw_text or "",
            "status": status_meta["status"],
            "error": status_meta["error"],
        })
        return opts, status_meta

    # Validate distractor lengths; replace any wildly off-length distractor
    validated = [opts[0]]
    fallback_count = 0
    for d in opts[1:]:
        ok, reason = passes_length_check(d, opts[0])
        if ok:
            validated.append(d)
        else:
            fallback_count += 1
            alt = semantically_distinct_negative_sample(opts[0], all_meanings, rng, n=1)[0]
            validated.append(alt)

    status = "generated" if fallback_count == 0 else ("partial" if fallback_count < 3 else "length_fallback")
    status_meta = {"status": status, "fallback_count": fallback_count}

    RAW_OUTPUTS.append({
        "stage": "generation",
        "variant": variant["name"],
        "language": language,
        "proverb": proverb,
        "correct_meaning": meaning,
        "raw_model_output": raw_text or "",
        "status": status_meta["status"],
        "fallback_count": status_meta.get("fallback_count", 0),
    })
    return validated, status_meta


# ── MCQ assembly ────────────────────────────────────────────────────────────

def assemble_mcq(opts, idx):
    """Place correct answer at deterministic rotating position."""
    correct_pos = idx % 4
    distractors = opts[1:]
    ordered = distractors[:correct_pos] + [opts[0]] + distractors[correct_pos:]
    labels = ["A", "B", "C", "D"]
    return {labels[i]: ordered[i] for i in range(4)}, labels[correct_pos]


# ── Blind shortcut audit ────────────────────────────────────────────────────

def extract_choice(text):
    """Extract a single A-D choice. Prefer the LAST occurrence to avoid
    capturing option letters that appear inside the model's reasoning text."""
    if not text:
        return None
    text = text.strip()
    # Strong signal: a standalone letter at the very start or end of the response.
    first_line = text.splitlines()[0].strip()
    if len(first_line) == 1 and first_line in "ABCD":
        return first_line
    last_line = text.splitlines()[-1].strip()
    if len(last_line) == 1 and last_line in "ABCD":
        return last_line
    # Otherwise take the last bounded A-D anywhere in the response.
    matches = re.findall(r"\b([A-D])\b", text.upper())
    return matches[-1] if matches else None


def audit_one_mcq(options_dict, correct_label, variant_name, mcq_id):
    text = "\n".join([f"{k}. {v}" for k, v in options_dict.items()])
    messages = [
        {"role": "system", "content": "Reply ONLY with A, B, C, or D."},
        {"role": "user", "content": f"Which option is correct?\n\n{text}"},
    ]
    prompt_tokens_est = sum(len(m["content"].split()) for m in messages) + 20

    votes = {}
    for model in AUDIT_COMMITTEE:
        cost_tracker.check()
        if cost_tracker.would_exceed(model, prompt_tokens_est, MAX_TOKENS_AUDIT):
            votes[model] = None
            continue
        try:
            resp = openrouter_chat(model, messages, MAX_TOKENS_AUDIT)
            vote_text = extract_text(resp)
            prompt_tok, out_tok = extract_usage(resp)
            cost_tracker.add(model, prompt_tok or prompt_tokens_est, out_tok, "audit")
            votes[model] = extract_choice(vote_text)
            RAW_OUTPUTS.append({
                "stage": "audit",
                "variant": variant_name,
                "mcq_id": mcq_id,
                "model": model,
                "prompt": messages[1]["content"],
                "raw_model_output": vote_text,
                "extracted_vote": votes[model],
            })
        except Exception as e:
            votes[model] = None
            RAW_OUTPUTS.append({
                "stage": "audit",
                "variant": variant_name,
                "mcq_id": mcq_id,
                "model": model,
                "prompt": messages[1]["content"],
                "raw_model_output": "",
                "extracted_vote": None,
                "error": str(e),
            })
    return votes


def compute_consensus(votes, correct_label):
    valid = [v for v in votes.values() if v is not None]
    if not valid:
        return None, 0.0, 0
    most_common = Counter(valid).most_common(1)[0]
    consensus_label, consensus_count = most_common
    consensus_frac = consensus_count / len(valid)
    consensus_correct = 1 if consensus_label == correct_label else 0
    return consensus_label, consensus_frac, consensus_correct


# ── Main pipeline ───────────────────────────────────────────────────────────

def main():
    print(f"[{datetime.now(timezone.utc).isoformat()}] Starting Pilot 1 v2")
    print(f"Generator: {GENERATOR_MODEL}")
    print(f"Audit committee: {AUDIT_COMMITTEE}")
    print(f"Prompt variants: {[v['name'] for v in PROMPT_VARIANTS]}")
    print(f"Cost cap: ${COST_CAP_USD}")

    # Load & sample
    df_all = load_data()
    df_sample = sample_per_language(df_all, N_PER_LANG, SEED)
    df_sample = df_sample.reset_index(drop=True)
    all_meanings = df_all["meaning"].dropna().unique().tolist()
    rng = random.Random(SEED)

    print(f"\nSampled {len(df_sample)} proverbs:")
    print(df_sample.groupby("language").size())

    # Generate MCQs for each prompt variant
    all_mcqs = []
    for variant in PROMPT_VARIANTS:
        print(f"\n=== Generating with prompt variant: {variant['name']} ===")
        for idx, row in df_sample.iterrows():
            try:
                opts, meta = generate_options(
                    row["proverb"], row["meaning"], row["language"], variant, rng, all_meanings
                )
            except Exception as e:
                print(f"  Generation failed for {variant['name']} {row['language']}-{idx}: {e}")
                opts = [row["meaning"]] + negative_sample(row["meaning"], all_meanings, rng)
                meta = {"status": "hard_fallback", "error": str(e)}

            options_dict, correct_label = assemble_mcq(opts, idx)
            all_mcqs.append({
                "variant": variant["name"],
                "mcq_id": f"{variant['name']}_{row['language']}_{idx}",
                "language": row["language"],
                "sample_id": row["sample_id"],
                "proverb": row["proverb"],
                "correct_meaning": row["meaning"],
                "correct_label": correct_label,
                "option_A": options_dict["A"],
                "option_B": options_dict["B"],
                "option_C": options_dict["C"],
                "option_D": options_dict["D"],
                "generation_status": meta.get("status", "unknown"),
                "fallback_count": meta.get("fallback_count", 0),
            })
        print(f"  Cumulative cost so far: ${cost_tracker.spent:.4f}")

    mcqs_df = pd.DataFrame(all_mcqs)

    # Blind shortcut audit
    print("\n=== Running blind shortcut audit ===")
    audit_records = []
    for _, row in mcqs_df.iterrows():
        options_dict = {
            "A": row["option_A"],
            "B": row["option_B"],
            "C": row["option_C"],
            "D": row["option_D"],
        }
        votes = audit_one_mcq(options_dict, row["correct_label"], row["variant"], row["mcq_id"])
        consensus_label, consensus_frac, consensus_correct = compute_consensus(
            votes, row["correct_label"]
        )
        record = {
            "variant": row["variant"],
            "mcq_id": row["mcq_id"],
            "language": row["language"],
            "correct_label": row["correct_label"],
            "consensus_label": consensus_label,
            "consensus_frac": consensus_frac,
            "consensus_correct": consensus_correct,
        }
        for model, vote in votes.items():
            record[f"vote_{model.replace('/', '_')}"] = vote
            record[f"hit_{model.replace('/', '_')}"] = int(vote == row["correct_label"]) if vote else None
        audit_records.append(record)
        if len(audit_records) % 10 == 0:
            print(f"  Audited {len(audit_records)}/{len(mcqs_df)} MCQs; cost: ${cost_tracker.spent:.4f}")

    audit_df = pd.DataFrame(audit_records)

    # Per-variant summary
    comparison_rows = []
    for variant_name in [v["name"] for v in PROMPT_VARIANTS]:
        sub = audit_df[audit_df["variant"] == variant_name]
        for lang in ["English", "Arabic", "Yoruba"]:
            lang_sub = sub[sub["language"] == lang]
            if len(lang_sub) == 0:
                continue
            comparison_rows.append({
                "variant": variant_name,
                "language": lang,
                "n": len(lang_sub),
                "consensus_accuracy": lang_sub["consensus_correct"].mean(),
                "mean_consensus_frac": lang_sub["consensus_frac"].mean(),
                "flag": flag_accuracy(lang_sub["consensus_correct"].mean()),
            })
        comparison_rows.append({
            "variant": variant_name,
            "language": "ALL",
            "n": len(sub),
            "consensus_accuracy": sub["consensus_correct"].mean(),
            "mean_consensus_frac": sub["consensus_frac"].mean(),
            "flag": flag_accuracy(sub["consensus_correct"].mean()),
        })

    comparison_df = pd.DataFrame(comparison_rows)

    # Per-model summary
    per_model_rows = []
    for variant_name in [v["name"] for v in PROMPT_VARIANTS]:
        sub = audit_df[audit_df["variant"] == variant_name]
        for model in AUDIT_COMMITTEE:
            col = f"hit_{model.replace('/', '_')}"
            valid = sub[col].notna()
            if valid.sum() == 0:
                continue
            per_model_rows.append({
                "variant": variant_name,
                "model": model,
                "accuracy": sub.loc[valid, col].mean(),
                "n": valid.sum(),
            })
    per_model_df = pd.DataFrame(per_model_rows)

    # Print results
    print("\n=== Per-variant consensus accuracy ===")
    print(comparison_df.pivot(index="variant", columns="language", values="consensus_accuracy").round(3).to_string())

    print("\n=== Per-model accuracy by variant ===")
    print(per_model_df.to_string(index=False))

    # Save outputs
    mcqs_path = os.path.join(OUTPUT_DIR, "pilot1_generated_mcqs.csv")
    audit_path = os.path.join(OUTPUT_DIR, "pilot1_audit_results.csv")
    comparison_path = os.path.join(OUTPUT_DIR, "pilot1_prompt_comparison.csv")
    raw_path = os.path.join(OUTPUT_DIR, "pilot1_raw_outputs.csv")
    summary_path = os.path.join(OUTPUT_DIR, "pilot1_summary.json")

    mcqs_df.to_csv(mcqs_path, index=False, encoding="utf-8-sig")
    audit_df.to_csv(audit_path, index=False, encoding="utf-8-sig")
    comparison_df.to_csv(comparison_path, index=False, encoding="utf-8-sig")
    if RAW_OUTPUTS:
        pd.DataFrame(RAW_OUTPUTS).to_csv(raw_path, index=False, encoding="utf-8-sig")

    summary = {
        "run_timestamp": datetime.now(timezone.utc).isoformat(),
        "seed": SEED,
        "n_per_language": N_PER_LANG,
        "generator_model": GENERATOR_MODEL,
        "audit_committee": AUDIT_COMMITTEE,
        "prompt_variants": [
            {"name": v["name"], "rationale": v["rationale"]} for v in PROMPT_VARIANTS
        ],
        "cost_cap_usd": COST_CAP_USD,
        "estimated_cost_usd": round(cost_tracker.spent, 6),
        "per_variant_comparison": comparison_df.to_dict(orient="records"),
        "per_model_accuracy": per_model_df.to_dict(orient="records"),
        "cost_history": cost_tracker.history,
    }
    with open(summary_path, "w", encoding="utf-8") as f:
        json.dump(summary, f, ensure_ascii=False, indent=2)

    print(f"\nSaved outputs to {OUTPUT_DIR}")
    print(f"Total estimated cost: ${cost_tracker.spent:.4f} / ${COST_CAP_USD:.2f}")


def flag_accuracy(acc):
    if pd.isna(acc):
        return "N/A"
    if acc < 0.30:
        return "PASS"
    elif acc <= 0.45:
        return "REVIEW"
    else:
        return "FAIL"


````
